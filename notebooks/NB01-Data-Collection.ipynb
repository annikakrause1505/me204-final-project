I started my data collection by setting Open Library's search API and creating an empty list to store all books pulled. I then loop through 2016-2025. I reset the page counter at th end of each year and limited it to 500 records per year. I used a while loop to request pages until the 5 page max is hit. I then used try: to send the HTTP get request. This part required outside research. I then pulled a list of records out of the response. I then add the records onto all.docs. Finally, I increase the page counter, add a 1 second wait to avoid overwhelming the API, and print my results to check for accuracy. 

```python
import requests
import time
import json

url = "https://openlibrary.org/search.json"
all_docs = []

for year in range(2016, 2026):
    print(f"--- Fetching year {year} ---")
    page = 1
    pages_per_year = 5  

    while page <= pages_per_year:
        params = {
            "q": f"first_publish_year:{year}",
            "subject": "fiction",
            "fields": "title,first_publish_year,number_of_pages_median,edition_key,key,author_name",
            "limit": 100,
            "page": page
        }

        try:
            resp = requests.get(url, params=params, timeout=10)
            resp.raise_for_status()
            data = resp.json()
        except (requests.exceptions.RequestException, json.JSONDecodeError) as e:
            print(f"  Error on year {year}, page {page}: {e}")
            time.sleep(5)
            continue

        docs = data.get("docs", [])
        if not docs:
            print(f"  No more results at page {page}")
            break

        all_docs.extend(docs)
        page += 1
        time.sleep(1)

    print(f"  Collected so far: {len(all_docs)}")

print(f"\nTotal collected: {len(all_docs)}")

with open("nb01_raw_fiction_2016_2025_TEST.json", "w") as f:
    json.dump(all_docs, f)
print("Saved sample.")
```

The next step in my data collection was creating the necessary file path, opening the file, and adds my all_docs list to the empty json file. I then print a confirmation to verify accuracy.

```python
import os

os.makedirs("data/raw", exist_ok=True)

with open("data/raw/openlibrary_fiction_2016_2025.json", "w") as f:
    json.dump(all_docs, f, indent=2)

print("Saved", len(all_docs), "records to data/raw/openlibrary_fiction_2016_2025.json")
```

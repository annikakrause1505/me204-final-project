```python
import requests
import time
import json

url = "https://openlibrary.org/search.json"
all_docs = []

for year in range(2016, 2026):
    print(f"--- Fetching year {year} ---")
    page = 1
    pages_per_year = 5  

    while page <= pages_per_year:
        params = {
            "q": f"first_publish_year:{year}",
            "subject": "fiction",
            "fields": "title,first_publish_year,number_of_pages_median,edition_key,key,author_name",
            "limit": 100,
            "page": page
        }

        try:
            resp = requests.get(url, params=params, timeout=10)
            resp.raise_for_status()
            data = resp.json()
        except (requests.exceptions.RequestException, json.JSONDecodeError) as e:
            print(f"  Error on year {year}, page {page}: {e}")
            time.sleep(5)
            continue

        docs = data.get("docs", [])
        if not docs:
            print(f"  No more results at page {page}")
            break

        all_docs.extend(docs)
        page += 1
        time.sleep(1)

    print(f"  Collected so far: {len(all_docs)}")

print(f"\nTotal collected: {len(all_docs)}")

with open("nb01_raw_fiction_2016_2025_TEST.json", "w") as f:
    json.dump(all_docs, f)
print("Saved sample.")
```

```python
import os

os.makedirs("data/raw", exist_ok=True)

with open("data/raw/openlibrary_fiction_2016_2025.json", "w") as f:
    json.dump(all_docs, f, indent=2)

print("Saved", len(all_docs), "records to data/raw/openlibrary_fiction_2016_2025.json")
```
